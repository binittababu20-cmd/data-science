```python
#6 Create a DataFrame containing the columns Name, Age, and Rating, and display the following DataFrame properties: Transpose of the DataFrame. Row and column labels using axes. Data type of each column using dtypes. Whether the DataFrame is empty using empty. Number of dimensions using ndim.Shape of the DataFrame using shape.Total number of elements using size.Values stored in the DataFrame using values. Number of missing values in each column using isnull().sum().


df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, 30, 35, 28, 22],
    'Rating': [4.5, 3.9, 4.2, 4.8, 3.7]
})
df.index = ['A', 'B', 'C', 'D', 'E']
print("DataFrame:")
display(df)
print("Transpose:")
display(df.T)
print("Row and column:")
display(df.axes)
print("Data type:")
display(df.dtypes)
print( df.empty)
print("Number of dimensions:", df.ndim)
print("Shape :", df.shape)
print("Total number of elements :", df.size)
print("Values stored in the DataFrame:")
display(df.values)
print("Number of missing values in each column :")
display(df.isnull().sum())

```
